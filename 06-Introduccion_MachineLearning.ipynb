## Diagramas de dipersión

Los diagramas de dispersión, también conocidos como gráficos de dispersión o scatter plots en inglés, son una herramienta de visualización de datos que se utiliza para mostrar la relación entre dos variables numéricas. Cada punto en el gráfico representa una observación en el conjunto de datos, con una variable asignada al eje X y la otra al eje Y.

Utilidades de los diagramas de dispersión:

**Identificar correlaciones:**

Los diagramas de dispersión son excelentes para identificar la relación entre dos variables. Si los puntos están cercanamente agrupados y forman un patrón, esto puede indicar una relación positiva o negativa entre las variables.

**Detectar tendencias:**

Puedes ver cómo los valores de una variable se comportan en relación con otra. Por ejemplo, en un estudio económico, podrías usar un diagrama de dispersión para ver cómo cambia el consumo a medida que aumenta el ingreso de los hogares.

**Visualizar agrupaciones y outliers:**

Los gráficos de dispersión permiten identificar rápidamente datos que no se ajustan a la tendencia general (outliers) o grupos de datos (clusters). Esto es útil en la detección de errores, anomalías o para identificar subgrupos dentro de los datos.

**Comparar conjuntos de datos:**

Al graficar múltiples conjuntos de datos en el mismo gráfico de dispersión, puedes comparar visualmente sus distribuciones y correlaciones.
**Modelado predictivo:**

En estadística y machine learning, los diagramas de dispersión pueden ayudar a evaluar la adecuación de un modelo para predecir una variable a partir de otra y para verificar supuestos lineales, si es pertinente.

#  Diagrama de cajas y bigotes

Los diagramas de cajas y bigotes, también conocidos como box plots, son una forma de representación gráfica que permite describir distribuciones de datos de manera compacta y visual. Esta técnica fue desarrollada por John Tukey en la década de 1970 como parte de su enfoque de análisis exploratorio de datos.

## Estructura de un diagrama de cajas y bigotes:

#### Caja:
Línea inferior (Q1): Representa el primer cuartil de la distribución, el 25% de los datos están por debajo de este valor.
Línea superior (Q3): Representa el tercer cuartil de la distribución, el 75% de los datos están por debajo de este valor.
Línea media (Mediana): Divide la caja y representa el segundo cuartil o mediana de la distribución, es decir, el punto medio de los datos.

#### Bigotes:
Se extienden desde la caja hasta el valor máximo y mínimo dentro de un rango aceptable, normalmente calculado como 1.5 veces el rango intercuartílico (IQR = Q3 - Q1) por encima de Q3 y por debajo de Q1. Datos fuera de este rango se consideran outliers y se pueden marcar con puntos o alguna otra forma gráfica.

## Utilidad de los diagramas de cajas y bigotes:

**Resumen visual de la distribución:**
Los diagramas de cajas y bigotes resumen la distribución de los datos, mostrando la mediana, los cuartiles y los extremos. Esto permite comparar rápidamente distintos conjuntos de datos.

**Identificación de simetría y sesgos en los datos:**
La posición de la mediana dentro de la caja puede indicar si los datos son simétricos o si hay un sesgo hacia uno de los extremos.

**Detección de valores atípicos (outliers):**
Los puntos que caen fuera del rango de los bigotes se consideran atípicos y son visualmente identificables. Esto es útil para identificar datos que podrían ser errores de medición o que requieren un análisis especial.

**Comparación entre grupos:**
Los diagramas de cajas y bigotes son especialmente útiles para comparar distribuciones entre varios grupos o condiciones en un experimento, haciendo fácil ver diferencias en mediana, dispersión y presencia de outliers entre grupos.

# Los Histrogramas

Los histogramas son una herramienta gráfica muy común en estadística para representar la distribución de un conjunto de datos numéricos. Al crear un histograma, los datos se dividen en intervalos o "bins", y se cuenta el número de observaciones que caen en cada uno de estos intervalos. El resultado se representa como un gráfico de barras, donde cada barra representa un intervalo y la altura de la barra indica la frecuencia de observaciones dentro de ese intervalo.

## Utilidad de los Histogramas:

**Visualizar la distribución de los datos:**

Los histogramas proporcionan una visualización clara de cómo están distribuidos los datos. Puedes ver rápidamente si los datos están sesgados, si son simétricos, si siguen una distribución normal, o si presentan picos en ciertos valores.

**Identificar la centralidad y la dispersión:**
Los histogramas ayudan a identificar dónde se concentran los valores (la centralidad) y cuán extendidos están (la dispersión o variabilidad).

**Detectar anomalías y outliers:**
Las anomalías en los datos, como los outliers, pueden identificarse fácilmente con un histograma, ya que aparecerán como barras aisladas en lugares inesperados.

**Facilitar la estimación de la densidad:**
Los histogramas son una forma de estimación de la densidad empírica; pueden usarse para estimar la función de densidad de probabilidad subyacente a los datos.

**Comparar conjuntos de datos:** Se pueden usar múltiples histogramas para comparar las distribuciones de diferentes conjuntos de datos en el mismo conjunto de ejes, lo cual es útil para comparaciones visuales directas.

```python
# Histogramas

import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Cargar datos de Iris desde scikit-learn
datosIris = load_iris()

# Crear una figura
plt.figure(figsize=(8, 6))  # Ajustar tamaño de la figura

# Título general
plt.suptitle('Histograma de los atributos del Iris', fontsize=16, fontweight='bold')

# Lista de nombres de características
features = datosIris['feature_names']

# Subplot 1 - Sepal Length
plt.subplot(2, 2, 1)
plt.hist(datosIris['data'][:, 0], bins=20, color='blue', alpha=0.7)
plt.title(features[0])  # Subtítulo

# Subplot 2 - Sepal Width
plt.subplot(2, 2, 2)
plt.hist(datosIris['data'][:, 1], bins=20, color='green', alpha=0.7)
plt.title(features[1])  # Subtítulo

# Subplot 3 - Petal Length
plt.subplot(2, 2, 3)
plt.hist(datosIris['data'][:, 2], bins=20, color='red', alpha=0.7)
plt.title(features[2])  # Subtítulo

# Subplot 4 - Petal Width
plt.subplot(2, 2, 4)
plt.hist(datosIris['data'][:, 3], bins=20, color='purple', alpha=0.7)
plt.title(features[3])  # Subtítulo

# Ajustar el diseño para evitar superposición
plt.tight_layout(rect=[0, 0, 1, 0.96])

# Mostrar gráfico
plt.show()
```

##  Otra vista de histogramas:
### Distribución de los atributos del Iris por clase: Histogramas con estimación de densidad
             

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.datasets import load_iris

# Cargar datos de Iris desde scikit-learn
datosIris = load_iris()

# Convertir a DataFrame
iris_df = pd.DataFrame(datosIris.data, columns=datosIris.feature_names)
iris_df['species'] = [datosIris.target_names[i] for i in datosIris.target]

# Lista de características
features = datosIris.feature_names
# Lista de especies
species = iris_df['species'].unique()
# Colores para cada especie
colors = {'setosa': 'red', 'versicolor': 'green', 'virginica': 'blue'}

# Crear la figura
plt.figure(figsize=(12, 10))

# Titulo de la imagen
plt.suptitle("Distribución de los atributos del Iris por clase: Histogramas con estimación de densidad",
             fontsize=14, fontweight='bold')
# Generar histogramas organizados por atributo
for i, feature in enumerate(features):
    plt.subplot(2, 2, i + 1)  # Organizar en una cuadrícula 2x2
    for specie in species:
        subset = iris_df[iris_df['species'] == specie][feature]
        # Dibujar el histograma con transparencia
        plt.hist(subset, bins=20, alpha=0.5, color=colors[specie], label=specie, density=True)
        # Dibujar la curva de densidad para suavizar la distribución
        sns.kdeplot(subset, color=colors[specie], linewidth=2)

    # Agregar título y etiquetas
    plt.title(f'Histograma de {feature.replace(" (cm)", "")}', fontsize=12, fontweight='bold')
    plt.xlabel("Valor")
    plt.ylabel("Densidad")
    plt.legend()

# Ajustar el diseño para evitar superposición
plt.tight_layout()
plt.show()
```

## Matriz de confusión

Una matriz de confusión es una herramienta utilizada en el campo del aprendizaje automático para evaluar el desempeño de un modelo de clasificación. Es una tabla que permite la visualización del rendimiento de un algoritmo, especialmente útil en problemas de clasificación supervisada.

Es una tabla con dos dimensiones: "Predicted" (Predicho) y "Actual" (Real), que compara los valores reales de los datos de prueba con los predichos por el modelo. Cada fila de la matriz representa las instancias en una clase real, mientras que cada columna representa las instancias en una clase predicha.

Para un problema de clasificación binaria, la matriz de confusión se ve como sigue:


### Componentes de la Matriz de Confusión

**Verdaderos Positivos (TP):** Casos en los que la predicción es positiva, y el valor real también es positivo.

**Falsos Negativos (FN):** Casos en los que la predicción es negativa, pero el valor real es positivo.

**Falsos Positivos (FP):** Casos en los que la predicción es positiva, pero el valor real es negativo.

**Verdaderos Negativos (TN):** Casos en los que la predicción es negativa, y el valor real también es negativo.


### Utilidad de la matriz de confusión

**Evaluación de Rendimiento:**
Permite una evaluación cuantitativa del desempeño del modelo. Proporciona una visión detallada no solo de los aciertos del modelo (TP y TN), sino también de los tipos de errores que está cometiendo (FP y FN).

**Cálculo de Métricas:**
A partir de la matriz de confusión, se pueden calcular varias métricas de rendimiento importantes, como:

- *Precisión:*
  ¿Qué proporción de identificaciones positivas fue realmente correcta?

- *Recuperación (Recall):*
  Qué proporción de casos positivos reales se identificó correctamente?

- *Puntuación F1:*
  Una medida que combina la precisión y el recall en una sola métrica, útil cuando las clases son desequilibradas.

- *Especificidad:*
  ¿Qué proporción de casos negativos reales se identificó correctamente?

**Diagnóstico de Problemas de Clasificación:**
Ayuda a identificar si un modelo está teniendo dificultades con un tipo particular de error, como predecir en exceso una clase o ser demasiado conservador. Es especialmente útil en problemas con clases desequilibradas o en situaciones donde ciertos tipos de errores tienen consecuencias diferentes.

**Mejora del Modelo:**
Al identificar los tipos de errores cometidos por un modelo, los desarrolladores pueden trabajar específicamente en mejorar el modelo en esos aspectos. Por ejemplo, ajustando el umbral de decisión o añadiendo más datos representativos para clases mal predichas.

```python
from sklearn.metrics import confusion_matrix

print("\n Matriz de confusión: \n")
skcm = confusion_matrix(y_prueba, y_pred)

# colocar en un dataframe para imprimir las etiquetas
skcm = pd.DataFrame(skcm, columns=['predice-setosa','predice-versicolor','predice-virginica'])
skcm['actual'] = ['setosa','versicolor','virginica']
skcm = skcm.set_index('actual')
print(skcm)
```

### Precisión, Recuperación y F1 Score

**Precisión**: Proporción de todas las predicciones positivas que son correctas. La precisión es una medida de cuántas predicciones positivas fueron observaciones positivas reales

$$Precisión = \frac{VP}{VP + FP} = \frac{positivos\ predichos\ correctamente}{todas\ las\ predicciones\ positivas}$$

La precisión es útil cuando el enfoque principal es minimizar los falsos positivos, es decir, cuando se desea asegurar que las predicciones positivas sean altamente confiables y no contengan demasiados falsos positivos.

**Recuperación**: Proporción de todas las observaciones positivas reales que son correctas. La recuperación es una medida de cuántas observaciones positivas reales se pronosticaron correctamente. (De todas las positivas, cuantas se pronosticaron)

$$Recuperación = \frac{VP}{VP + FN} = \frac{predichos\ de\ ser\ positivos}{todas\ las\ observaciones\ positivas}$$

La recuperación es especialmente importante cuando el enfoque principal es minimizar los falsos negativos, es decir, cuando se desea asegurar que se detecten la mayor cantidad posible de casos positivos reales.

**F1 Score**,  Métrica que tiene en cuenta la precisión y la recuperación. Es la media armónica de estas 2 métricas y se calcula como tal:

$$F1 = 2\times\frac{precisión \times recuperación}{precisión + recuperación}$$

El F1-score proporciona una medida general del rendimiento del modelo, teniendo en cuenta tanto los falsos positivos como los falsos negativos. Es especialmente útil cuando se busca un equilibrio entre la precisión y la recuperación.

```python
from sklearn.metrics import classification_report

print("\nReporte de la Clasificación:")
print(classification_report(y_prueba, y_pred, target_names=['setosa', 'versicolor', 'virginica']))

# Precision: qué proporción de los detectados positivos son realmente positivos
# Recuperación (recall): Qué proporcion de los positivosreales, se detectaron.
```

# Validación cruzada

La validación cruzada es una técnica de evaluación de modelos en estadísticas y aprendizaje automático que se utiliza para evaluar la capacidad de generalización de un modelo a un conjunto de datos independiente. Es especialmente útil cuando se busca evitar el sobreajuste, asegurando que el modelo no solo funcione bien en los datos de entrenamiento, sino también en nuevos datos no vistos.

Consiste en dividir el conjunto de datos en varias partes (subconjuntos) y realizar iteraciones en las que cada subconjunto es utilizado una vez como datos de prueba y las demás veces como datos de entrenamiento. Las más comunes son la validación cruzada de k particiones y la validación cruzada de Leave-One-Out (LOO).

## Tipos de Validación Cruzada

### K-Fold Cross-Validation:
- El conjunto de datos se divide en "k" subconjuntos.
- El modelo se entrena en "k-1" de estos subconjuntos y se prueba en el subconjunto restante.
- Este proceso se repite "k" veces, con cada uno de los "k" subconjuntos utilizado exactamente una vez como conjunto de prueba.
- La precisión del modelo se promedia sobre las "k" pruebas para dar una medida de rendimiento.

### Leave-One-Out (LOO):
- Es un caso especial de la validación cruzada de k particiones donde "k" es igual al número de observaciones en el conjunto de datos.
- En cada iteración, se deja una muestra fuera como el conjunto de prueba, y el resto se utiliza como conjunto de entrenamiento.
- Este proceso se repite tantas veces como muestras haya en el conjunto de datos.

## Propósitos de la Validación Cruzada

 - **Estimación de la Capacidad de Generalización:**
     Al usar diferentes subconjuntos de datos para entrenamiento y prueba, ayuda a evitar el problema del sobreajuste y proporciona una estimación fiable de la capacidad de generalización del modelo sobre nuevos datos.

 - **Selección de Modelos:**
     Permite comparar la eficacia de diferentes configuraciones de modelos o hiperparámetros, eligiendo aquel que tenga el mejor rendimiento  promedio en las diferentes iteraciones de la validación cruzada.

  - **Evaluación Robusta:**
      Al promediar los resultados de múltiples pruebas, la validación cruzada proporciona una medida más robusta y menos sesgada del rendimiento del modelo en comparación con una simple división de entrenamiento/prueba.
      
## Fundamento Matemático que Sustenta su Efectividad

La efectividad de la validación cruzada está fundamentada en la teoría estadística de la estimación y la inferencia. Básicamente, al utilizar múltiples divisiones de datos, la validación cruzada minimiza los sesgos y las varianzas asociadas con una única partición de los datos de entrenamiento y prueba.

- **Reducción del Sesgo:**
    Al rotar los subconjuntos de datos utilizados para la prueba, se asegura que todos los datos se utilizan tanto para entrenamiento como para pruebas, lo que ayuda a evitar sesgos en la estimación del rendimiento del modelo.

- **Estabilidad de la Varianza:**
    El promedio de múltiples estimaciones tiene una varianza menor que una única estimación. La validación cruzada, al promediar los resultados de los modelos entrenados en diferentes subconjuntos, tiende a tener una varianza más baja que una división única de entrenamiento/prueba.
