##
Análisis de la Base de Datos Wine

La base de datos Wine (Vino) proviene de un análisis químico de vinos cultivados en la región de Italia, derivados de tres variedades diferentes de uvas. Cada registro contiene mediciones de 13 atributos químicos y físicos (por ejemplo, contenido de alcohol, ácido málico, magnesio, etc.), con el objetivo de clasificar el vino en una de las tres variedades (0, 1, 2). En total, se incluyen 178 muestras de vino, lo que la hace útil para estudios de clasificación y análisis exploratorio de datos.

Para dar contexto al caso, describimos las variedades de vino a que hace referencia el dataset:

###  Variedades de vino en el *Wine Dataset*

|  **Clase** |  **Variedad de uva** |  **Descripción** |
|:-------------:|:----------------------|:-------------------|
| **0** | **Barolo** | Vino tinto robusto, con alta concentración de taninos y acidez marcada. Ideal para guarda prolongada. |
| **1** | **Grignolino** | Vino más ligero y aromático, de color más claro y menor contenido de taninos. Fresco y afrutado. |
| **2** | **Barbera** | Vino de cuerpo medio, color intenso y alta acidez, típico del Piamonte. Muy equilibrado y versátil. |
